## check duplicate

```python
data.duplicated().sum()
```

## drop duplicate

```python
data.drop_duplicates(inplace=True)
```

## check null value

```python
data.isnull().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.shape
```

## replace unit price from object to float

```python
data['unit_price'] = data['unit_price'].str.replace('$', '').astype(float)  
data.head() 
```

```python
data.info()
```

```python
data["Total"] = data["quantity"] * data["unit_price"]
data.head()
```

```python
data.columns
```

## replace all column to lower case

```python
data.columns = data.columns.str.strip().str.lower().str.replace(' ', '_')
data.head()                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                         
```
